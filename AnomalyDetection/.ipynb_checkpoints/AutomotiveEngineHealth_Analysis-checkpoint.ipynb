## Visualize Data distribution

```python
# Target Varible distribution

plt.figure(figsize = (6,5))

counts = data["Engine Condition"].value_counts()
percentages = counts / counts.sum() * 100

# Plot bar chart

bars = plt.bar(
    counts.index,
    counts.values,
    color=['lightcoral', 'lightblue'],
    edgecolor='black'
)

# Add titles and labels
plt.title("Distribution of Engine Condition", fontsize=14, fontweight='bold')
plt.xlabel("Condition Category", fontsize=12)
plt.ylabel("Frequency", fontsize=12)

# Annotate with percentages
for bar, pct in zip(bars, percentages):
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f"{pct:.1f}%",
        ha='center',
        va='bottom',
        fontsize=11,
        fontweight='bold'
    )

# Improve layout
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

The target variable is slightly imbalanced. This can affect the classification model performance, especially in metrices like precision, recall and F1-score for the minority class.
Classifiers like Logistics regression, Random Forest etc. optimize for Overall accuracy. This can lead to underperformance with minority class.
Chances of poor precision or recall for the minority class
Selection of decision threshold probability will be incorrect

How to overcome this:
- Evaluate using Confusion Matrix, precision recall curves and F1 Score
- Use stratified sampling during train-test
- Use resampling techniques like:
  - SMOTE (Synthetic Minority Over-sampling Technique)
  -  Random undersampling of the majority class
